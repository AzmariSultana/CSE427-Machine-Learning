# Matplotlib

```python
import matplotlib.pyplot as plt
```

##Task1

```python
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]

plt.scatter(temperature_dhaka, humidity_dhaka, color='red', marker='*')

plt.xlabel('Temperature')
plt.ylabel('Humidity')
plt.title('Temperature vs Humidity')
plt.show()
```

##Task2

```python
study_hours = [2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]

# line plot, fig size(12,8)
plt.figure(figsize=(12,8))
plt.plot(study_hours, marks, color='red', marker='o')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks')
plt.show()
```

##Task3

```python
subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks = [89, 90, 45, 78, 99]
plt.barh(subjects, marks, color='pink')
plt.xlabel('Marks')
plt.ylabel('Subjects')
plt.title('Subjects vs Marks')
plt.show()
```

##Task4

```python
colors = ['brown', 'lavender', 'skyblue', 'orange', 'purple']
plt.bar(subjects, marks, color=colors)
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Subjects vs Marks')
plt.show()
```

## Task5

```python
study_marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
subject_marks = [89, 90, 45, 78, 99]

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.scatter(temperature_dhaka, humidity_dhaka, color='red', marker='*')
plt.xlabel('Temperature')
plt.ylabel('Humidity')
plt.title('Temperature vs Humidity (Dhaka)')
plt.grid()

plt.subplot(2, 2, 2)
plt.plot(study_hours, study_marks, color='red', marker='o')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks')
plt.grid()

plt.subplot(2, 2, 3)
plt.barh(subjects, subject_marks, color='pink')
plt.xlabel('Marks')
plt.ylabel('Subjects')
plt.title('Subjects vs Marks')

plt.subplot(2, 2, 4)
plt.bar(subjects, subject_marks, color=colors)
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Subjects vs Marks')

plt.tight_layout()
plt.show()
```
